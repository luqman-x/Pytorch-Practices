```python
from sklearn.model_selection import GroupShuffleSplit
import numpy as np
import matplotlib.pyplot as plt

# Example dataset
X = np.random.rand(100, 5)
y = np.random.randint(0, 2, 100)

# 10 groups, each containing 10 samples
groups = np.array([i // 10 for i in range(100)])

# GroupShuffleSplit
gss = GroupShuffleSplit(
    n_splits=1,
    test_size=0.2,
    random_state=42
)

train_idx, test_idx = next(
    gss.split(X, y, groups=groups)
)

X_train, X_test = X[train_idx], X[test_idx]
y_train, y_test = y[train_idx], y[test_idx]


# Visualization
# Assign a label to each sample
split = np.full(100, "Train", dtype=object)
split[test_idx] = "Test"

# Plot
plt.figure(figsize=(12, 4))

for i in range(100):
    if split[i] == "Train":
        plt.scatter(i, groups[i], marker="s", s=100)
    else:
        plt.scatter(i, groups[i], marker="s", s=100)

# Add group boundaries
for g in range(1, 10):
    plt.axhline(g - 0.5, alpha=0.3)

plt.xlabel("Sample Index")
plt.ylabel("Group")
plt.title("GroupShuffleSplit: Train/Test Assignment")

plt.yticks(range(10))
plt.grid(axis="x", alpha=0.2)

plt.show()
```
